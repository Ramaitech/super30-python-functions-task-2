```python
"""Return the largest value among all supplied positional arguments."""

def largest(*args):
    largest_number = args[0]

    for number in args:
        if number > largest_number:
            largest_number = number

    return largest_number


print(largest(10, 25, 15))
print(largest(5, 100, 25, 50))
print(largest(12, 45, 7, 89, 23, 67))
```

```python
"""
Function Practice: *args
------------------------
This module demonstrates how to use *args to accept any number of
positional arguments. Two functions are shown:

1. total(*args)    -> returns the sum of all supplied values
2. largest(*args)  -> returns the largest supplied value
"""


def total(*args):
    """
    Return the total (sum) of all supplied positional arguments.

    Parameters:
        *args: A variable-length list of numeric values.

    Returns:
        The sum of all values in args.
    """
    # sum(args) is the clean, Pythonic way to add all values
    return sum(args)


def largest(*args):
    """
    Return the largest value among all supplied positional arguments.

    Parameters:
        *args: A variable-length list of numeric values.

    Returns:
        The largest value in args.
    """
    # Start by assuming the first value is the largest
    largest_number = args[0]

    # Compare each number with the current largest
    for number in args:
        if number > largest_number:
            largest_number = number

    return largest_number


# -----------------------------
# Sample calls for total(*args)
# -----------------------------
print("total(10, 20, 30):", total(10, 20, 30))          # 60
print("total(5, 15):", total(5, 15))                    # 20
print("total(1, 2, 3, 4, 5):", total(1, 2, 3, 4, 5))    # 15

# -----------------------------
# Sample calls for largest(*args)
# -----------------------------
print("largest(10, 25, 15):", largest(10, 25, 15))      # 25
print("largest(5, 100, 25, 50):", largest(5, 100, 25, 50))  # 100
```
