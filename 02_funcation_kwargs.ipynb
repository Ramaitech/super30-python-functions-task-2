```python
#def create_profile(**kwargs)

def create_profile(**kwargs):
    profile = {}
    for key, value in kwargs.items():
        profile[key] = value
    return profile

print(create_profile(name="Alice", age=30, city="New York"))
print(create_profile(username="bob123", email="bob@example.com"))
print(create_profile(first_name="Charlie", last_name="Brown", country="USA"))
```

```python
# def Function that accepts another funcaiton
def add(a,b):
    return a+b
def multiply(a,b):
    return a*b
def calculator(func, a, b):
    return func(a, b)

print(calculator(add, 10, 20))
print(calculator(multiply, 10, 20))




```
