```python
#def create_profile(**kwargs)

def create_profile(**kwargs):
    profile = {}
    for key, value in kwargs.items():
        profile[key] = value
    return profile

print(create_profile(name="Alice", age=30, city="New York"))
print(create_profile(username="bob123", email="bob@example.com"))
print(create_profile(first_name="Charlie", last_name="Brown", country="USA"))
```

```python
# def Function that accepts another funcaiton
def add(a,b):
    return a+b
def multiply(a,b):
    return a*b
def calculator(func, a, b):
    return func(a, b)

print(calculator(add, 10, 20))
print(calculator(multiply, 10, 20))




```

```python
def create_profile(**kwargs):
    """
    Print all provided user attributes and also return them as a dictionary.

    Parameters:
        **kwargs: Arbitrary keyword arguments representing user attributes.

    Behavior:
        - Prints each key-value pair in a clean, readable format.
        - Returns the kwargs dictionary for optional further use.
    """

    print("User Profile Information:")
    
    # Loop through all key-value pairs and print them
    for key, value in kwargs.items():
        print(f"{key}: {value}")

    # Returning the dictionary is optional but useful for testing or reuse
    return kwargs


# -----------------------------
# Sample calls
# -----------------------------

create_profile(name="Rahul", age=22, city="Bangalore")

create_profile(username="yjd123", role="Admin", active=True, score=98)
```
