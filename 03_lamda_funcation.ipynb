```python
# Lambda function to calculate square
square = lambda x: x**2 # Lambda function to calculate the square of a number
print(square(10)) # Output: 100
```

```python
# Create a lambda function for calculating the square of a number.
numbers = [1, 2, 3, 4, 5, 6]

# map() applies the lambda function to each element in the list
# lambda x: x * x means "take x and return x squared"
squared_numbers = list(map(lambda x: x * x, numbers))

print(squared_numbers)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

# filter() keeps only the elements where the lambda returns True
# lambda x: x % 2 == 0 checks if x is even
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

```python



# Recursive function to calculate factorial of a number
def factorial(n):
    # Base case: factorial of 0 or 1 is 1
    if n == 0 or n == 1:
        return 1
    
    # Recursive case: n * factorial of (n-1)
    return n * factorial(n - 1)


# Sample calls
print(factorial(5))   #
print(factorial(6))   # 720
```

```python
"""
Recursion practice:
- factorial(n)
- sum_to_n(n)      -> 1 + 2 + ... + n
- fibonacci(n)     -> nth Fibonacci number
"""


def factorial(n: int) -> int:
    """Return n! using recursion."""
    if n == 0 or n == 1:          # Base case
        return 1
    return n * factorial(n - 1)   # Recursive case


def sum_to_n(n: int) -> int:
    """Return 1 + 2 + ... + n using recursion."""
    if n == 1:                    # Base case
        return 1
    return n + sum_to_n(n - 1)    # Recursive case


def fibonacci(n: int) -> int:
    """
    Return the nth Fibonacci number using recursion.
    Sequence: 0, 1, 1, 2, 3, 5, 8, ...
    """
    if n == 0:                    # Base case 1
        return 0
    if n == 1:                    # Base case 2
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)  # Recursive case


# Sample calls
print("factorial(5):", factorial(5))       # 120
print("sum_to_n(5):", sum_to_n(5))         # 15  (1+2+3+4+5)
print("fibonacci(7):", fibonacci(7))       # 13  (0,1,1,2,3,5,8,13)
```

```python
"""
Recursion Practice Functions
----------------------------
This file implements three required recursive functions:

1. factorial(n)   -> returns n!
2. sum_to_n(n)    -> returns 1 + 2 + ... + n
3. fibonacci(n)   -> returns the nth Fibonacci number

Each function includes:
- A clear base case
- A recursive case
- Sample calls for demonstration
"""


def factorial(n: int) -> int:
    """
    Compute n! (factorial of n) using recursion.

    Base cases:
        factorial(0) = 1
        factorial(1) = 1

    Recursive rule:
        n! = n * (n - 1)!
    """
    if n == 0 or n == 1:  # Base case
        return 1
    return n * factorial(n - 1)  # Recursive case


def sum_to_n(n: int) -> int:
    """
    Compute the sum 1 + 2 + ... + n using recursion.

    Base case:
        sum_to_n(1) = 1

    Recursive rule:
        sum_to_n(n) = n + sum_to_n(n - 1)
    """
    if n == 1:  # Base case
        return 1
    return n + sum_to_n(n - 1)  # Recursive case


def fibonacci(n: int) -> int:
    """
    Compute the nth Fibonacci number using recursion.

    Fibonacci sequence:
        0, 1, 1, 2, 3, 5, 8, ...

    Base cases:
        fibonacci(0) = 0
        fibonacci(1) = 1

    Recursive rule:
        fibonacci(n) = fibonacci(n - 1) + fibonacci(n - 2)
    """
    if n == 0:  # Base case 1
        return 0
    if n == 1:  # Base case 2
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)  # Recursive case


# -----------------------------
# Sample calls for verification
# -----------------------------

print("factorial(5):", factorial(5))       # Expected: 120
print("sum_to_n(5):", sum_to_n(5))         # Expected: 15
print("fibonacci(7):", fibonacci(7))       # Expected: 13
```
